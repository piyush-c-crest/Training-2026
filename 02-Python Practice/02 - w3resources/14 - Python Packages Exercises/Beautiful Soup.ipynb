1. Write a Python program to find the title tags from a given html document.

```python
import requests
import bs4
```

```python
url = 'https://quotes.toscrape.com'
```

```python
res = requests.get(url)
res
```

```python
soup = bs4.BeautifulSoup(res.text,'lxml')
soup
```

```python
soup.select('title')
```

```python
soup.select('p')
```

```python
len(soup.select('p'))
```

```python
len(soup.find_all('p'))
```

```python
soup.select('p')[0]
```
