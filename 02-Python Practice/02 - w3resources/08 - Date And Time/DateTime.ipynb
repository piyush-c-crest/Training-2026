```python
print(date.year)
print(date.month)
print(date.weekday())
print(date.day)
```

- %Y: Year with century (e.g., 2026)
- %y: Year without century (e.g., 26)
- %m: Month as a zero-padded decimal (e.g., 01)
- %B: Full month name (e.g., January)
- %b: Abbreviated month name (e.g., Jan)
- %d: Day of the month as a zero-padded decimal (e.g., 01)
- %A: Full weekday name (e.g., Friday)
- %a: Abbreviated weekday name (e.g., Fri)
- %H: Hour (24-hour clock, zero-padded) (e.g., 13)
- %I: Hour (12-hour clock, zero-padded) (e.g., 01)
- %M: Minute as a zero-padded decimal (e.g., 04)
- %S: Second as a zero-padded decimal (e.g., 12)
- %p: Locale's equivalent of either AM or PM (e.g., PM)
- %c: Locale's appropriate date and time representation (e.g., Fri Jan 16 13:04:12 2026)
- %%: A literal % character 

```python
date = datetime.date.today()
```

```python
print(date.strftime('%Y'))
print(date.strftime('%B'))
print(date.strftime('%W'))
print(date.strftime('%A'))
print(date.strftime('%W'))
print(date.strftime('%j'))
print(date.strftime('%d'))
```
