### 1. Maximum of Three Numbers

Write a Python function to find the maximum of three numbers.

```python
def foo(a,b,c):
    if a > b:
        if a > c:
            return a
        return c
    elif b > c:
        return b
    return c
```

```python
print(foo(1,2,3))
print(foo(1,3,2))
print(foo(3,1,2))
```

### 1. Iterator from Iterables

Write a Python program to create an iterator from several iterables in a sequence and display the type and elements of the new iterator.

```python
lst = [1,2,3]
i = iter(lst)
```

```python
next(i)
```

```python
next(i)
```

```python
from itertools import chain
def chain_func(l1,l2,l3):
    return chain(l1,l2,l3)  
    
#List
result = chain_func([1,2,3], ['a','b','c','d'], [4,5,6,7,8,9])
print("Type of the new iterator:")
print(type(result))
print("Elements of the new iterator:")
for i in result:
    print(i)

#Tuple
result = chain_func((10,20,30,40), ('A','B','C','D'), (40,50,60,70,80,90))
print("Type of the new iterator:")
print(type(result))
print("Elements of the new iterator:")
for i in result:
    print(i)
```

### 1. Even Number Filter

Write a Python function that filters out even numbers from a list of integers using the filter function.

```python
list(filter(lambda x : x%2 == 0, [1,2,3,4,5,6]))
```
