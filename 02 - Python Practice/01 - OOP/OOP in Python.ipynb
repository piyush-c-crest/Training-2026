```python
class Human():
    def __init__(self,name, age):
        self.name = name
        self.age = age

    def compare(self, c):
        if self.age == c.age:
            return True
        else: return False
```

```python
p1 = Human('A', 20)
p2 = Human('B', 20)
```

```python
if p1.compare(p2):
    print('same age')
else:
    print('not same age')
```

```python
p2.age = 22
```

```python
if p1.compare(p2):
    print('same age')
else:
    print('not same age')
```

## Inner Class

```python
class Student():
    def __init__(self, name):
        self.name = name
        self.lap = self.Laptop()
        
    class Laptop(): # INNER CLASS
        def __init__(self):
            self.brand = 'HP'
            self.ram = 16
        def show(self):
            print(self.brand, self.ram)
    
```

```python
lap = Student.Laptop()
```

```python
lap.show()
```

## Inheritance

```python
class A:
    def f1(self):
        pass
    def f2(self):
        pass
        
```

```python
class B(A):
    def f3(self):
        pass
```

```python
b = B()
```

```python
b.f3()
```

```python
b.f1() # acees A's methods
```

```python
class D(A,C):  #  multiple inheritance
    pass
```

## Polymorphism

### Duck typing

```python
class Pycharm:
    def execute(s):
        print('Compiling')
        print('Running')
```

```python
class Jupyter:
    def execute(s):
        print('Debugging')
        print('Compiling')
        print('Running')
```

```python
class A:
    def code(s,ide): # Duck typing
        ide.execute()
```

```python
a = A()
```

```python
a.code(Pycharm())
```

```python
a.code(Jupyter())
```

```python
a = 5
b = 6
a+b
```

```python
class Student:
    def __init__(self,m1,m2):
        self.m1 = m1
        self.m2 = m2

    def __add__(self,other): # magic Method
        s3 = Student(self.m1+other.m1, self.m2+other.m2)
        return s3
    def __gt__(self,other): # magic Method
        return (self.m1+self.m2 >  other.m1+other.m2)
```

```python
s1 = Student(10,20)
s2 = Student(50,50)
```

```python
s3 = s1+s2
```

```python
s3.m1, s3.m2
```

```python
if s1 > s2:
    print('S1 wins')
else: 
    print('S2 wins')
```
