## **Exercise 2: Adding and Subtracting Matrices**

### **Problem Statement**

Write a Python program to **add and subtract two matrices** of the same dimensions.

### **Tasks**

1. Create two matrices of equal dimensions.
2. Perform matrix addition.
3. Perform matrix subtraction.
4. Display the results clearly.



```python
l1 = [[1,2,3],[3,4,5],[6,7,8]]
l2 = [[1] * 3] * 3
l2
```

```python
for i in range(len(l1)):
    for j in range(len(l1[i])):
        l1[i][j] += l2[i][j]

l1
```

## **Exercise 3: Matrix Multiplication**

### **Problem Statement**

Write a Python program to multiply two matrices.

### **Tasks**

1. Accept two matrices.
2. Validate whether multiplication is possible.
3. Multiply the matrices using nested loops.
4. Store the result in a new matrix.
5. Display the resulting matrix.

```python
a = [[1,2],[3,4]]
b = [[2,1],[4,3]]

res = [[0,0],[0,0]]
```

```python
for i in range(len(a)):
    for j in range(len(b[0])):
        #result[i][j] = 0
        for k in range(len(b)):
            res[i][j] += a[i][k] * b[k][j]
```

```python
res
```
