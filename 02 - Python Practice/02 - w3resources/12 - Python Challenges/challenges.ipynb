4. Write a Python program to check if a number is a perfect square.
Input : 9
Output : True

```python
def is_perfect_square(num):
    sqr = int(num**0.5)
    return sqr*sqr == num
```

```python
is_perfect_square(13)
```

```python
is_perfect_square(16)
```

5. Write a Python program to check if an integer is the power of another integer.
Input : 16, 2
Output : True


```python
import math
```

```python
def foo(n1,n2):
    return 'n1 is power of n2' if math.log(n1,n2).is_integer() else 'not a power'
```

```python
foo(16,2)
```

```python
foo(17,2)
```

```python
print(foo(16, 2))
print(foo(12, 2))
print(foo(81, 3))
```

12. Write a Python program to find the single number in a list that doesn't occur twice.
Input : [5, 3, 4, 3, 4]
Output : 5

```python
l = [5, 3, 4, 3, 4] 
for i in set(l):
    if l.count(i) == 1:
        print(i)
```
