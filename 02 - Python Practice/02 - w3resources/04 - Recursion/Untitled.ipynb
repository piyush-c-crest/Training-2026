### 1. Sum of List Using Recursion

Write a Python program to calculate the sum of a list of numbers using recursion.

```python
lst = [1,2,3,4,5,6]
```

```python
def foo(lst,index=0):
    if index == len(lst):
        return 0
    else:
        return lst[index] + foo(lst,index+1)
```

```python
foo(lst)
```

```python
sum(lst)
```
