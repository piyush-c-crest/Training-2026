3. SQLite Table Creation

Write a Python program to connect a database and create a SQLite table within the database.

```python
try:
    connection = sqlite3.connect('mydb.db')
    con = connection.cursor()
    query = 'create table emp( id char(25), name char(25), phone char(10))'
    con.execute(query)
    print('created successfully')
    con.close()
except:
    print('Somethong went wrong')
finally:
    connection.close()
```

```python
try:
    connection = sqlite3.connect('mydb40.db')
    con = connection.cursor()
    query = 'create table emp( id char(25), name char(25), phone char(10))'
    con.execute(query)
    print('created successfully')
    query = 'insert into emp values ("100", "A", "1010101010"); insert into emp values("101", "B", "1010101011")'
    con.executescript(query)
    print('inserted successfully')
    conn.commit()
    con.execute("SELECT * FROM emp")
    rows = con.fetchall()
    print("emp details:")
    for row in rows:
       print(row)
    con.close()
except:
    print('Somethong went wrong')
finally:
    connection.close()
```
