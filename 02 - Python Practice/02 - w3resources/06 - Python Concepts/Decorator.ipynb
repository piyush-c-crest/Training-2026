### 1. Create a Decorator to Log Function Arguments and Return Value

Write a Python program to create a decorator that logs the arguments and return value of a function.

```python
def new_deco(fun):

    def logger(*arg, **kwarg):
        print('arguments : ' ,arg)
        print('Keyword arguments : ' ,kwarg)
        res = fun(*arg)
        print('result : ' ,res)
        return res
        
    return logger
```

```python
@new_deco
def fun(*arg):
    return sum(arg)
```

```python
fun(1,2)
```
