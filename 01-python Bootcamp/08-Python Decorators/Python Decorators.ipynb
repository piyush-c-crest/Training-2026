```python
def hello(name):
    print('inside Hello')

    def greet():
        print('hello '+ name)
    def welcome():
        print('Welcome to Python')

    greet()
    welcome()
```

```python
def hello(name=None):
    def greet():
        print('hello '+ name)
    def welcome():
        print('Welcome to Python')

    if type(name) == type(''):
        return greet
    else:
        return welcome
```

```python
hello()()
```

```python
hello('Piyush')()
```

## Decorators

```python
def new_decorator(og_fun):

    def wrap_fun():
        print('Some extra code before og fun')
        og_fun()
        print('some extra code after og fun')

    return wrap_fun
```

```python
def fun():
    print('fun which get decorated')
```

```python
var = new_decorator(fun)
var()
```

### syntax for decorator

```python
@new_decorator
def fun():
    print('fun which get decorated')
```

```python
fun()
```
