```python
path = 'C:\\Users\\PiyushChunara\\Desktop\\Training-2026\\01-python Bootcamp'
for dirpath, dirnames, files in os.walk(path): # walk returns ->  dirpath (string), dirnames (list of string), filenames (list of string)
    if '.' in dirpath:
        continue
        
    print('current directory is '+ dirpath.split('\\')[-1])

    print('\n directories in folder : ')
    for dir in dirnames:
        print('\t'+dir)
        
    print('\n Files in folder : ')
        
    for file in files:
        print('\t'+file)
    print('*'*100)
```

## Regular expression

```python
import re
```

```python
text = 'hello how are you'
pattern = 'how'
```

```python
re.search(pattern,text)
```

```python
re.search('pattern',text)
```

```python
match = re.search(pattern,text)
```

```python
match.span()
```

```python
text = 'hello how are  how you'
```

```python
match = re.findall(pattern,text)
```

```python
match
```

```python
for i in re.finditer(pattern, text):
    print(i)
```
