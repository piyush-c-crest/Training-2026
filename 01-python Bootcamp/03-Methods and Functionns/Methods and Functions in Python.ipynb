### guess the ball game

```python
from random import shuffle
```

```python
def shuf_list(l):
    shuffle(l)
    return l
```

```python
l = [' ', ' ', 'O']
shuf_list(l)
```

```python
def player_guess():
    ip = ''
    while ip not in ['0','1','2']:
        ip = input('Guess btwn 0,1,2 : ')

    return int(ip)    
```

```python
player_guess()
```

```python
def check_guess(l,guess):
    if l[guess] == 'O':
        print('Correct')
        return
    print('Wrong')
    print(l)
```

```python
# full code

mylist = [' ', ' ', 'O']

mixed = shuf_list(mylist)

guess = player_guess()

check_guess(mixed,guess)
```

## **arg and *kwargs
- arg (argument) -> tuple
- kwargs (key word argument) -> dictionary

```python
def get_sum(*arg):
    """Calculates the sum of a variable number of arguments."""
    return sum(arg)
print(get_sum(1,2,3))
print(get_sum(10,20,30,40,50))
```

```python
def foo(*a):
    print('data type : ' , type(a))
    print(a)

foo(1,2,3)
foo(1,2,3,4,5)
```

```python
def intro(**data):
    print("Data type : ", type(data))
    print("Data : ", data)
    for key, value in data.items():
        print(f"{key}: {value}")

intro(fname="Piyush", lname="Chunara", age=20, city="Anand")
```

```python
def fun(*arg , **kwargs):
    print(arg)
    print(kwargs)
    print(arg[0], kwargs['a'])

fun(1,2,3,5, a='p',b='q', )
```
