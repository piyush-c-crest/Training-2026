## Map fun in py

```python
nums = [1,2,3,4]
```

```python
def square(x):
    return x**2
```

```python
map(square,nums)
```

```python
for i in map(square,nums):
    print(i)
```

```python
list(map(square,nums))
```

## Filter Function
- filter(fun,obj)
- fun -> which checks item , must take argument and must return bollean (true or false)
- obj -> iterable like list , tuple

```python
def is_even(i):
    return i%2==0
```

```python
filter(is_even, [1,2,3,4,5,6,7,])
```

```python
list(filter(is_even, [1,2,3,4,5,6,7,]))
```

```python
for i in filter(is_even, [1,2,3,4,5,6,7,]):
    print(i)
```

## Lambda Function

```python
def sqrt(n):
    res = n ** 2
    return res
```

```python
sqrt(9)
```

```python
s = lambda n : n ** 2
```

```python
s(4)
```

```python
list(map(lambda x : x*2, [1,2,3,4]))
```

```python
list(filter(lambda x : x%2, [1,2,3,4]))
```

## Scope of Variable

Local > Enclosing > Global > Builtin 

```python
x = 'I am Global var'

def f1():
    x = 'I am Enclosing'
    def f2():
        x = 'i am Local'
        print(x)
    f2()

f1()
```

```python
x = 'Global'
def foo():
    x = 'New Val'
    print(x)
```

```python
foo()
```

```python
print(x)
```

```python
x = 'Global'
def foo():
    global x
    x = 'New Val'
    print(x)
```

```python
foo()
```
