## Generators
- In Python, generators are special functions that allow you to create iterators in a simple and memory-efficient way. Unlike normal functions that return a single value and terminate, generators can pause execution and yield multiple values one at a time, on demand. 
Key Concepts
- **yield keyword**: This is what makes a function a generator. When yield is encountered, the function's state (local variables, execution point, etc.) is saved, the value is returned to the caller, and the function pauses.
- **Lazy Evaluation**: Generators produce value
- s only when requested (e.g., by a for loop or the next() function), instead of computing all values and storing them in memory at once.
- **Generator Object**: When you call a generator function, it doesn't run the code immediately; instead, it returns a generator object, which is an iterator. 

```python
def create_cube(n):
    for i in range(n):
        yield i**3
```

```python
create_cube
```

```python
create_cube(10)
```

```python
for i in create_cube(10):
    print(i)
```

```python
def fib(n):
    a,b = 1,1
    for i in range(n):
        print('--')
        yield a
        print('**')
        a,b = b, a+b
        
```

```python
for i in fib(10):
    print(i)
```
