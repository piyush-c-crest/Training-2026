- r -> read only
- w -> write only (will overwrite or creates nrew file)
- a -> append only (will add new lines to exiting file)
- r+ -> read + write 
- w+ -> same

```python
with open('x.txt', mode='r') as x:
    print(x.read())
```

```python
with open('x.txt', mode='a') as x:
    x.write('Hello')
```

```python
with open('x.txt', mode='r') as x:
    print(x.read())
```
