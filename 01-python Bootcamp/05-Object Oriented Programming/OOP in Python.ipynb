```python
class Dog():
    # will be same for all instace of a class like static
    species = 'mamal'
    def __init__(self, breed, name, spots):
        self.breed = breed
        self.name = name
        self.spots = spots

    # Methods 
    def bark(self):
        print('WOOF!')

    def foo(self, num):
        print(f'name is: {self.name} , number is : {num}')
```

```python
dog = Dog('Lab','a','No')
dog.bark()
```

```python
dog.foo(10)
```

## inheritance and polymorphism

```python
class Animal():
    def __init__(self):
        print('I am animal')

    def foo():
        print('hi from animal')
        
    def eat(self):
        print('eating')
```

```python
class Dog(Animal):  # inheritance

    def __init__(self):
        Animal.__init__(self)
        print('I am Dog')
        
    def foo(self): #overwrite
        print('hi from dog')
```

```python
d = Dog()
```

```python
d.eat()
```

```python
d.foo()
```

```python
class Dog():
    def __init__(self):
        print('Dog')
    def speak(self):
        print('Dog speaks woof')
```

```python
class Cat():
    def __init__(self):
        print('Cat')
    def speak(self):
        print('Cat speaks meow')
```

```python
def pet_speak(pet):
    pet.speak()
```

```python
dog = Dog()
cat = Cat()
```

```python
pet_speak(dog)
pet_speak(cat)
```

## Special Functions

```python
class Book():
    def __init__(self, title, name):
        self.title = title
        self.name = name

    def __str__(self):
        return f"{self.title} by {self.name}"

    def __len__(self):
        return 100

    def __cc__(self):
        print('Nothing')
        
```

```python
b = Book('start with Python', 'XYZ')
print(b)
```
