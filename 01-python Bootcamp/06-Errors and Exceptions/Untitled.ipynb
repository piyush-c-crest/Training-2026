## Errors and Exception Handling

```python
try:
    # code which can have error
    res = 10 + 10 + '10'
except:
    # this code execute if error occured
    print('cant add string with int')
else:
    # executes if works fine
    print('No Error')
    print(res)
finally:
    print('final block')
```

```python
try:
    with open('text.txt', 'w') as file:
        file.write('hello')
except TypeError: # to handel specific type of errors only
    print('Type Error Occured')
except OSError:
    print('OS Error Occured')
finally:
    print('I always run!')
```

```python
def ask_ip():
    while 1:
        try:
            res = int(input('Enter number : '))
        except :
            print('enter a number plz')
            continue
        else:
            print('yes correct')
            break
        finally:
            print('end of code')
```
