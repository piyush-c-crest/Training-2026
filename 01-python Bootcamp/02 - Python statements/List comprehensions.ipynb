## List comprehensions
### used to make for loop smaller

```python
str = 'hello'
```

```python
new = []
for i in str:
    new.append(i)
new
```

```python
new = [i for i in str]
new
```

```python
[i**2 for i in range(0,11)]
```

```python
[i**2 for i in range(0,11) if i%2==0]
```

```python
cls = [0,10,30.5,60,5]
fer = [((9/5)*c + 32) for c in cls]
```

```python
fer
```

```python
[x if x%2==0 else 'odd' for x in range(11)]
```

### nested loop

```python
new = []
for i in [1,2,3]:
    for j in [10,20,30]: # first it will run for 1 , then 2 then 3
        new.append(i*j)
new
```

```python
x = [1,2,3]
[[i * j for j in [10,20,30,]] for i in x]
```

```python
[i * j for i in x for j in [10,20,30,]]
```
