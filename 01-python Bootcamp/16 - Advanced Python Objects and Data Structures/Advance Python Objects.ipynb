```python
s = set([1,2,3])
```

```python
s
```

```python
s.clear()
```

```python
s = {1,2,3}
sc = s.copy()
```

```python
sc.add(4) # adding in copy will not affect original
```

```python
sc, s
```

```python
sc.difference(s)
```
