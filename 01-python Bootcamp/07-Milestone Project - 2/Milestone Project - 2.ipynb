```python
suits = ('Hearts', 'Diamonds', 'Spades', 'Clubs')
ranks = ('Two', 'Three', 'Four', 'Five', 'Six', 'Seven', 'Eight', 'Nine', 'Ten', 'Jack', 'Queen', 'King', 'Ace')
values = {'Two':2, 'Three':3, 'Four':4, 'Five':5, 'Six':6, 'Seven':7, 'Eight':8, 
            'Nine':9, 'Ten':10, 'Jack':11, 'Queen':12, 'King':13, 'Ace':14}
```

```python
class Card:
    def __init__(self, suit, rank):
        self.suit = suit
        self.rank = rank
        self.value = values[rank]

    def __str__(self):
        return self.rank + " of " + self.suit
```

deck class
- creates all 52 cards
- shuffles cards (random.shuffle)


```python
class Deck():
    def __init__(self):
        self.all_cards = []
        for suit in suits:
            for rank in ranks:
                self.all_cards.append(Card(suit,rank)) #creating Card objects and stores in list

    def shuffle(self):
        random.shuffle(self.all_cards)

    def deal_one(self):
        return self.all_cards.pop()
```

class Player
- holds all cards
- can remove card from top -> pop(0)
- can add card or cards to bottom -> append or extend()
- 

```python
class Player():
    def __init__(self,name):
        self.name = name
        self.all_cards = []

    def remove_one(self):
        return self.all_cards.pop()

    def add_cards(self, new_cards):
        if type(new_cards) == type([]): # to check if input is 
        # list of cards or single card which is Card obj
           self.all_cards.extend(new_cards)
        else: 
            self.all_cards.append(new_cards)

    def __str__(self):
        return f'Player {self.name} have {len(self.all_cards)} cards.'

    def __len__(self):
        return len(self.all_cards)
```

```python
# game setup
p1 = Player('one')
p2 = Player('two')

new_deck = Deck()
new_deck.shuffle()

for i in range(26):
    p1.add_cards(new_deck.deal_one())
    p2.add_cards(new_deck.deal_one())
```

```python
game_on = True
round_num = 0
while game_on:
    
    round_num+=1
    print('at round ', round_num)
    
    if len(p1) == 0:
        game_on=False
        print('Player-1 is out of cards! Player-2 wins!')
        break
    if len(p2) == 0:
        game_on=False
        print('Player-2 is out of cards! Player-1 wins!')
        break

    p1_cards = []
    p2_cards = []

    p1_cards.append(p1.remove_one())
    p2_cards.append(p2.remove_one())

    at_war = True
    while at_war: 
        if p1_cards[-1].value > p2_cards[-1].value:
            p1.add_cards(p1_cards)
            p1.add_cards(p2_cards)
            at_war = False
            
        elif p1_cards[-1].value < p2_cards[-1].value:
            p2.add_cards(p1_cards)
            p2.add_cards(p2_cards)
            at_war = False

        else:
            print('WAR!!')

            if len(p1) < 3:
                print('Player one has no cards!')
                print('Player 2 Wins')
                game_on = False
                break
            elif len(p2) < 3:
                print('Player two has no cards!')
                print('Player 1 Wins')
                game_on = False
                break
            else:
                for i in range(3):
                    p1_cards.append(p1.remove_one())
                    p2_cards.append(p2.remove_one())
```
