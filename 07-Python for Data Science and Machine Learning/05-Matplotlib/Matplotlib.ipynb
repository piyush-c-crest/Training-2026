```python
import matplotlib.pyplot as plt
```

```python
import numpy as np
```

```python
x = np.linspace(1,10,10)
y = x ** 2
```

```python
plt.plot(x,y,'r--')
plt.xlabel('X lable')
plt.ylabel('Y lable')
plt.title('Plot')
```

```python
plt.subplot(1,2,1)
plt.plot(x,y,'r')

plt.subplot(1,2,2)
plt.plot(y,x,'g')
```

## Object Oriented Method

```python
fig = plt.figure()
axes = fig.add_axes([0.1,0.1,0.8,0.8])
axes.plot(x,y)
```

```python
fig = plt.figure()
axes1 = fig.add_axes([0.1,0.1,0.8,0.8])
axes2 = fig.add_axes([0.2,0.5,0.4,0.2])

axes1.plot(x,y)
axes2.plot(y,x)
```

```python
fig, axes = plt.subplots(nrows=2,ncols=2, figsize=(10,7))
axes[0,0].plot(x,y,'r')
axes[0,1].plot(x,y,'--')
axes[1,0].plot(x,y,'b.')
axes[1,1].plot(x,y,'^')
```

```python
fig.savefig('plot.pdf',dpi=400)
```

```python
plt.plot(x,x**2, label = 'Square')
plt.plot(x,x**3, label = 'cube')
plt.legend()
```

```python
plt.plot(x,x**2, label = 'Square', color='red', linewidth=5, alpha=0.2, linestyle='-.')
```

```python
plt.plot(x,y,marker='o', markerfacecolor='red', markersize=20)
```
