# Linear REgression

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df1 = pd.read_csv('USA_Housing.csv')
```

```python
df1.head()
```

```python
df = df1.copy()
```

```python
X = df.drop(columns=['Address', 'Price'])
y = df['Price']
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.33, random_state=102)
```

```python
from sklearn.linear_model import  LinearRegression
```

```python
lr = LinearRegression()
```

```python
lr.fit(X_train, y_train)
```

```python
lr.intercept_
```

```python
lr.coef_
```

```python
pd.DataFrame({'predict' : lr.predict(X_test),'actual' :y_test})
```

```python
pre = lr.predict(X_test)
```

```python
plt.scatter(y_test, pre)
plt.xlabel('Predicted')
plt.ylabel('Actual')
```

```python
from sklearn import metrics
```

```python
metrics.mean_absolute_error(y_test, pre)
```

```python
metrics.mean_squared_error(y_test, pre)
```

```python
np.sqrt(metrics.mean_squared_error(y_test, pre))
```
