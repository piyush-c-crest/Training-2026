## numpy indexing and selection

```python
arr = np.arange(0,10)
arr
```

```python
arr[3]
```

```python
arr[1:3]
```

```python
arr[:]
```

```python
arr[1:3] = 100
```

```python
arr
```

```python
arr = np.arange(1,10).reshape(3,3)
arr
```

```python
arr[1][2]
```

```python
arr[1,2]
```

```python
arr[:2,1:]
```

```python
arr[[1,2,]]
```

```python
arr = np.arange(1,11)
arr
```

```python
arr[arr > 5] 
```
